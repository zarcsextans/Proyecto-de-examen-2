```python
class Graph:
    def __init__(self):
        # Inicializa un diccionario vacío para representar la lista de adyacencia.
        # Las llaves serán los vértices y los valores serán listas de sus vecinos.
        self.adj = {}

    def add_vertex(self, v):
        # Agrega un nuevo vértice al grafo.
        # Si el vértice ya existe, no hace nada para evitar sobrescribir datos.
        if v not in self.adj:
            self.adj[v] = []

    def add_edge(self, u, v):
        # Agrega una arista entre los nodos 'u' y 'v'.
        # Primero asegura que ambos nodos existan en el diccionario.
        if u not in self.adj:
            self.adj[u] = []
        if v not in self.adj:
            self.adj[v] = []

        # Al ser un grafo no dirigido, la relación es bidireccional.
        # Agrega 'v' a la lista de 'u' y 'u' a la lista de 'v' si no están ya conectados.
        if v not in self.adj[u]:
            self.adj[u].append(v)
        if u not in self.adj[v]:
            self.adj[v].append(u)

    def neighbors(self, v):
        # Retorna la lista de nodos conectados directamente al vértice 'v'.
        return self.adj[v]

    def vertices(self):
        # Retorna una lista con todos los nombres/identificadores de los vértices.
        return list(self.adj.keys())

    def order_by_degree_desc(self):
        # Implementa una heurística: ordena los vértices de mayor a menor grado.
        # El "grado" es la cantidad de vecinos (longitud de su lista de adyacencia).
        return sorted(self.vertices(), key=lambda x: len(self.adj[x]), reverse=True)

    def greedy_coloring(self, order=None):
        # Si no se proporciona un orden específico, se usa el orden natural (alfabético/numérico).
        if order is None:
            order = sorted(self.vertices())

        # Diccionario para almacenar la asignación final {vértice: color}.
        coloring = {}

        # Itera sobre cada vértice según el orden definido.
        for v in order:
            # Conjunto para identificar qué colores ya están siendo usados por los vecinos.
            used_colors = set()
            for neighbor in self.neighbors(v):
                if neighbor in coloring:
                    used_colors.add(coloring[neighbor])

            # Busca el primer color (entero empezando desde 0) que no esté en used_colors.
            color = 0
            while color in used_colors:
                color += 1

            # Asigna el color disponible más bajo encontrado (comportamiento voraz o greedy).
            coloring[v] = color

        return coloring

    def validate_coloring(self, coloring):
        # Recorre todos los vértices y sus vecinos para verificar conflictos.
        for v in self.adj:
            for neighbor in self.adj[v]:
                # Si dos nodos conectados tienen el mismo color, la coloración es incorrecta.
                if coloring[v] == coloring[neighbor]:
                    return "No válida"
        # Si termina el ciclo sin encontrar conflictos, la solución es matemáticamente correcta.
        return "Válida"
```

```python
# 1. Instanciamos un objeto de la clase Graph
g1 = Graph()

# 2. Definimos la lista de vértices para el grafo completo
vertices = ["A", "B", "C", "D"]

# 3. Construcción del Grafo Completo K4
# Utilizamos un ciclo anidado para conectar cada nodo con todos los que le siguen en la lista
for i in range(len(vertices)):
    for j in range(i + 1, len(vertices)):
        # Agrega una arista entre cada par de nodos único
        g1.add_edge(vertices[i], vertices[j])

# 4. Aplicamos el algoritmo de coloración voraz (Greedy)
coloring_k4 = g1.greedy_coloring()

# 5. Despliegue de resultados en consola
print("K4")
for v in coloring_k4:
    # Muestra la asignación: Vértice > Color (0, 1, 2 o 3)
    print(v, ">", coloring_k4[v])

# 6. Cálculo del total de colores únicos utilizando un conjunto (set)
print("Total colores usados:", len(set(coloring_k4.values())))

# 7. Ejecución de la función de validación para asegurar que no haya conflictos
print("Validación:", g1.validate_coloring(coloring_k4))
```

```python
# 1. Instanciamos un segundo objeto de la clase Graph para la prueba del ciclo
g2 = Graph()

# 2. Definimos los vértices que formarán el anillo
cycle = ["A", "B", "C", "D", "E"]

# 3. Construcción del Grafo Ciclo C5
# El uso del operador módulo (%) es clave aquí: conecta cada nodo con el siguiente,
# y al llegar al último (E), lo conecta de regreso al primero (A) para cerrar el círculo.
for i in range(len(cycle)):
    g2.add_edge(cycle[i], cycle[(i + 1) % 5])

# 4. Ejecutamos el coloreo voraz sobre la estructura circular
coloring_c5 = g2.greedy_coloring()

# 5. Despliegue de la asignación de colores
print("C5")
for v in coloring_c5:
    print(v, ">", coloring_c5[v])

# 6. Verificamos la eficiencia: en un C5, el resultado esperado es 3 colores
print("Total colores usados:", len(set(coloring_c5.values())))

# 7. Comprobación de que la solución respeta las reglas del grafo
print("Validación:", g2.validate_coloring(coloring_c5))
```

```python
# 1. Ejecución del algoritmo usando el Orden Natural (A, B, C...)
# Se fuerza el orden alfabético para ver el comportamiento base del algoritmo.
coloreo_natural = g2.greedy_coloring(order=sorted(g2.vertices()))
n_natural = len(set(coloreo_natural.values()))

# 2. Ejecución del algoritmo usando la Heurística de Grado Descendente
# Aquí se utiliza el método 'order_by_degree_desc' que definimos en la clase Graph.
coloreo_grado = g2.greedy_coloring(order=g2.order_by_degree_desc())
n_grado = len(set(coloreo_grado.values()))

# 3. Impresión de resultados comparativos
# Esto permite analizar si el orden influyó en la cantidad de colores (número cromático).
print(f"Colores usados (Orden Natural): {n_natural}")
print(f"Colores usados (Grado Descendente): {n_grado}")

# 4. Validación final del método optimizado
# Aseguramos que, independientemente del orden, la solución sigue siendo válida.
print(f"Validación Grado Descendente: {g2.validate_coloring(coloreo_grado)}")
```
