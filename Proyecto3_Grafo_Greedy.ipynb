```python
class Graph:
    def __init__(self):

        self.adj = {}

    def add_vertex(self, v):
        # vertice
        if v not in self.adj:
            self.adj[v] = []

    def add_edge(self, u, v):
        # arista
        if u not in self.adj:
            self.adj[u] = []
        if v not in self.adj:
            self.adj[v] = []

        # grafo no dirigido
        if v not in self.adj[u]:
            self.adj[u].append(v)
        if u not in self.adj[v]:
            self.adj[v].append(u)

    def neighbors(self, v):
        # lista de nodos retornados
        return self.adj[v]

    def vertices(self):
        # lista de nodos
        return list(self.adj.keys())

    def order_by_degree_desc(self):
        # orden menor a mayor
        return sorted(self.vertices(), key=lambda x: len(self.adj[x]), reverse=True)

    def greedy_coloring(self, order=None):
        # Si no se proporciona un orden específico, se usa el orden natural
        if order is None:
            order = sorted(self.vertices())

        coloring = {}

        # Itera sobre cada vértice según el orden definido.
        for v in order:
            # colores usados por los vecinos
            used_colors = set()
            for neighbor in self.neighbors(v):
                if neighbor in coloring:
                    used_colors.add(coloring[neighbor])

            color = 0
            while color in used_colors:
                color += 1


            coloring[v] = color

        return coloring
# Validación de coloración
    def validate_coloring(self, coloring):

        for v in self.adj:
            for neighbor in self.adj[v]:

                if coloring[v] == coloring[neighbor]:
                    return "No válida"

        return "Válida"
```

```python
# Instanciamos
g1 = Graph()

# vertices
vertices = ["A", "B", "C", "D"]

# Construcción del Grafo Completo K4
for i in range(len(vertices)):
    for j in range(i + 1, len(vertices)):

        g1.add_edge(vertices[i], vertices[j])

# Aplicamos el algoritmo de coloración voraz (Greedy)
coloring_k4 = g1.greedy_coloring()

print("K4")
for v in coloring_k4:

    print(v, ">", coloring_k4[v])

print("Total colores usados:", len(set(coloring_k4.values())))

print("Validación:", g1.validate_coloring(coloring_k4))
```

```python
# Instancia
g2 = Graph()

# vertices
cycle = ["A", "B", "C", "D", "E"]

# 3. Construcción del Grafo Ciclo C5

for i in range(len(cycle)):
    g2.add_edge(cycle[i], cycle[(i + 1) % 5])

# coloreo
coloring_c5 = g2.greedy_coloring()

# asignación de colores
print("C5")
for v in coloring_c5:
    print(v, ">", coloring_c5[v])


print("Total colores usados:", len(set(coloring_c5.values())))


print("Validación:", g2.validate_coloring(coloring_c5))
```

```python

coloreo_natural = g2.greedy_coloring(order=sorted(g2.vertices()))
n_natural = len(set(coloreo_natural.values()))

#método 'order_by_degree_desc
coloreo_grado = g2.greedy_coloring(order=g2.order_by_degree_desc())
n_grado = len(set(coloreo_grado.values()))

print(f"Colores usados (Orden Natural): {n_natural}")
print(f"Colores usados (Grado Descendente): {n_grado}")

print(f"Validación Grado Descendente: {g2.validate_coloring(coloreo_grado)}")
```
